```python
import pandas as pd
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("spam (1).csv",encoding = "Latin_1")
```

Why encoding="Latin_1"?

The CSV file contains characters that cannot be decoded using UTF-8.
Therefore, Latin_1 encoding is used to read the dataset successfully without encountering a UnicodeDecodeError.

```python
df1["Label"] = df1["Label"].map({"ham":0,"spam":1})
```

```python
x = df1["Text"]
```

```python
y = df1["Label"]
```

```python
print(x.shape)
```

```python
print(y.shape)
```

```python
print(df1.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorise = TfidfVectorizer()
```

TF-IDF converts the text messages into numerical features that can be processed by the Machine Learning model.

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

The dataset is divided into 80% training dat

```python
x_train = vectorise.fit_transform(x_train)
```

```python
print(x_train.shape)
```

```python
x_test = vectorise.transform(x_test)
```

```python
print(x_test.shape)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model = MultinomialNB()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix,classification_report,f1_score
```

```python
print("Accuracy Score:",accuracy_score(y_test,y_pred))
```

```python
print("Recall Score:",recall_score(y_test,y_pred))
```

Recall:72% - The model correctlyidentifies 72% of actual spam messages. The lower recall indicates that some spam messages are misclassified as ham.

```python
print("Precision Score:",precision_score(y_test,y_pred))
```

```python
print("f1 score:",f1_score(y_test,y_pred))
```

```python
print("This is a confusion matrix:")
print(confusion_matrix(y_test,y_pred))
```

```python
print("CR:",classification_report(y_test,y_pred))
```
